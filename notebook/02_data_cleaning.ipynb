```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/data1.csv")
print("Dataset shape:", df.shape)
```

```python
target = "diseases"
X = df.drop(columns = target)
y = df[target]
print("Features:", X.shape)
print("Target:", y.shape)
```

```python
missing_values = df.isnull().sum().sum()
print("Total Missing values:", missing_values)
```

```python
unique_values = set(X.values.flatten())
print("Unique symptoms values:", unique_values)
```

```python
non_binary_columns = [col for col in X.columns if not set(X[col].unique()).issubset({0,1})]
print("Non-Binary columns:", non_binary_columns)
```

```python
before = len(df)
df = df.drop_duplicates()
after = len(df)
print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

```python
print("New Dataset shape:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
X = df.drop(columns = target)
pattern_disease_count = (
    df.groupby(X.columns.tolist())[target].nunique()
)
ambiguous_patterns = pattern_disease_count[pattern_disease_count > 1]
print("Ambiguous symptom patterns:", len(ambiguous_patterns))
```

```python
total_patterns = len(pattern_disease_count)
ambiguous_percentage = (len(ambiguous_patterns) / total_patterns) * 100
print(f"Total unique symptom patterns: {total_patterns}")
print(f"Ambiguous patterns: {len(ambiguous_patterns)}")
print(f"Ambiguous percentage: {ambiguous_percentage:.2f}%")
```

```python
print("Number of diseases:", df[target].nunique())
```

```python
disease_counts = df[target].value_counts()
print("Minimum samples:", disease_counts.min())
print("Maximum samples:", disease_counts.max())
print("Median samples:", disease_counts.median())
```

```python
disease_symptom_pattern = (
    df.groupby(target)[X.columns.tolist()].first()
)
duplicate_disease_profiles = disease_symptom_pattern[
    disease_symptom_pattern.duplicated(keep=False)
]
print("Diseases involved in identical profiles:", len(duplicate_disease_profiles))
```

```python
print("Final Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Features:", df.shape[1] - 1)
print("Diseases:", df[target].nunique())
```

```python
df.to_csv("../data/processed/disease_symptom_cleaned.csv", index = False)
print("cleaned dataset saved successfully")
```
